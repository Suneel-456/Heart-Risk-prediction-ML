# Heart Disease Prediction - EDA
Run from VS Code with the `.venv` kernel selected. `load_raw()` flips the source target so **1 = disease**.

```
import sys; sys.path.insert(0, '..')
import pandas as pd, seaborn as sns, matplotlib.pyplot as plt
from src.data_preprocessing import load_raw
df = load_raw()
df.shape, df.isna().sum().sum(), df.duplicated().sum()
```

```
df.describe().T
```

```
sns.countplot(x='target', data=df); plt.title('Class balance (1 = disease)'); plt.show()
```

```
fig, ax = plt.subplots(2, 3, figsize=(14, 7))
for a, col in zip(ax.ravel(), ['age','trestbps','chol','thalach','oldpeak']):
    sns.histplot(data=df, x=col, hue='target', kde=True, ax=a)
ax.ravel()[-1].axis('off'); plt.tight_layout(); plt.show()
```

```
plt.figure(figsize=(9,7)); sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm'); plt.show()
```

```
for col in ['cp','sex','exang','thal','ca','slope']:
    print(df.groupby(col)['target'].agg(['mean','count']).round(2), '\n')
```
